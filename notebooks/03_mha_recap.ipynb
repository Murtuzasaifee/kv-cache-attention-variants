# Session 3 — Multi-Head Attention Recap (standalone baseline)

From-scratch MHA with an incremental KV cache — the reference every later variant is compared against.

```python
import sys; sys.path.insert(0, '../src')
import torch
from kv_cache_variants.attention.mha import MultiHeadAttention

torch.manual_seed(0)
mha = MultiHeadAttention(d_model=32, num_heads=4)
x = torch.randn(1, 5, 32)
out, kv = mha(x)
print('output shape:', out.shape, '| cached K shape:', kv[0].shape)
```

```python
# Incremental decode step reusing the cache
next_x = torch.randn(1, 1, 32)
out2, kv2 = mha(next_x, past_kv=kv)
print('next-step output shape:', out2.shape, '| grown cache K shape:', kv2[0].shape)
```
