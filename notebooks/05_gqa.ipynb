# Session 5 — Grouped-Query Attention

GQA generalizes MHA (`num_kv_groups=num_heads`) and MQA (`num_kv_groups=1`).

```python
import sys; sys.path.insert(0, '../src')
import torch
from kv_cache_variants.attention.gqa import GroupedQueryAttention
from kv_cache_variants.memory_calc import kv_cache_bytes, human_bytes

torch.manual_seed(0)
d_model, num_heads = 32, 8
x = torch.randn(1, 5, d_model)
for groups in (8, 4, 2, 1):
    _, kv = GroupedQueryAttention(d_model, num_heads, groups)(x)
    b = kv_cache_bytes(num_layers=32, num_kv_heads=groups, head_dim=d_model // num_heads, seq_len=4096)
    print(f'num_kv_groups={groups}: cached heads={kv[0].shape[1]}, cache={human_bytes(b)}')
```
